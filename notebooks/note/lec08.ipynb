# Qubitization and Quantum Signal Processing

**Learning objectives.**

-   Derive Chebyshev transformations from a $2\times2$ rotation.

-   Qubitize Hermitian and general block encodings.

-   State the quantum signal processing (QSP) theorem.

-   Survey methods for finding phase factors.

The block-encoding calculus of that chapter let us add and multiply matrices,
and hence realize any *polynomial* of a block-encoded matrix $A$ as a short
linear combination of unitaries---at the price of an $\ell^1$ overhead that grows
with the degree. This lecture develops the far more economical route. The key
observation, due to Low and Chuang, is that a single block encoding $U_A$ of a
Hermitian $A$ is secretly a direct sum of $2\times2$ rotations, one per eigenvalue:
each eigenvector $\ket{v_j}$ spans, together with a companion "perpendicular"
state, a two-dimensional invariant subspace on which $U_A$ acts as a fixed
rotation by angle $\arccos\lambda_j$. This is *qubitization*
(that section). Once the eigenvalues sit inside independent qubits, powers of
the (reflected) block encoding compute Chebyshev polynomials
$T_k(\lambda_j)$ exactly (that section), and interleaving single-qubit
$Z$-rotations between the block encodings realizes an essentially arbitrary bounded
polynomial $P(A)$ with degree equal to the number of queries and subnormalization
independent of the degree. The precise class of achievable polynomials, and the
phase factors that select them, is the content of *quantum signal processing*
(that section). We treat the Hermitian and general (non-Hermitian) block
encodings separately (that section), because the alternation of
$U_A$ and $U_A\dg$ forced by a non-Hermitian encoding is exactly what makes the
even/odd parity distinction of the QSP theorem appear. We close
(that section) with a survey of how the phase factors are computed in
practice---a problem once thought hard, now routinely solved. Throughout, $A$ is an
$n$-qubit Hermitian matrix with $\norm{A}\le1$, block-encoded as in
that chapter; the singular-value generalization to non-Hermitian $A$ (QSVT) is
the subject of that chapter.

## Chebyshev polynomials from a $2\times2$ block

Everything in this lecture rests on one elementary fact about rotations. Recall the
*Chebyshev polynomials of the first and second kind*, defined for
$\lambda=\cos\theta$ by
$$T_k(\lambda)=\cos(k\theta)=\cos\!\bigl(k\arccos\lambda\bigr),\qquad
  U_{k-1}(\lambda)=\frac{\sin(k\theta)}{\sin\theta}
   =\frac{\sin\!\bigl(k\arccos\lambda\bigr)}{\sqrt{1-\lambda^2}}.$$
They satisfy the three-term recurrences
$$T_0=1,\quad T_1(\lambda)=\lambda,\quad T_{k+1}(\lambda)=2\lambda T_k(\lambda)-T_{k-1}(\lambda),$$
and the same recurrence holds for $U_k$ with $U_0=1$, $U_1=2\lambda$. In particular
$T_k$ has parity $k\bmod2$ and $\abs{T_k(\lambda)}\le1$ on $[-1,1]$.

**Lemma (Chebyshev polynomials from a rotation).**

Fix $-1\le\lambda\le1$, write $\lambda=\cos\theta$ with $0\le\theta\le\pi$, and set
$$U(\lambda)=\begin{pmatrix}\lambda&\sqrt{1-\lambda^2}\\[2pt]
    \sqrt{1-\lambda^2}&-\lambda\end{pmatrix},\qquad
  Z=\begin{pmatrix}1&0\\0&-1\end{pmatrix},\qquad
  O(\lambda)\defeq U(\lambda)Z.$$
Then $O(\lambda)$ is the planar rotation by $\theta$,
$$O(\lambda)=\begin{pmatrix}\lambda&-\sqrt{1-\lambda^2}\\[2pt]
    \sqrt{1-\lambda^2}&\lambda\end{pmatrix}
  =\begin{pmatrix}\cos\theta&-\sin\theta\\ \sin\theta&\cos\theta\end{pmatrix},$$
and its powers carry Chebyshev polynomials in every entry:
$$O(\lambda)^k
  =\begin{pmatrix}\cos(k\theta)&-\sin(k\theta)\\ \sin(k\theta)&\cos(k\theta)\end{pmatrix}
  =\begin{pmatrix}T_k(\lambda)&-\sqrt{1-\lambda^2}\,U_{k-1}(\lambda)\\[2pt]
    \sqrt{1-\lambda^2}\,U_{k-1}(\lambda)&T_k(\lambda)\end{pmatrix}.$$

*Proof.*

The matrix $U(\lambda)$ is a reflection ($U(\lambda)=U(\lambda)\dg$ and
$U(\lambda)^2=\id$ since $\lambda^2+(1-\lambda^2)=1$), and right-multiplication by
$Z$ negates the second column, giving the equation above; the identification with
$\bigl(\begin{smallmatrix}\cos\theta&-\sin\theta\\ \sin\theta&\cos\theta
\end{smallmatrix}\bigr)$ is immediate from $\lambda=\cos\theta$,
$\sqrt{1-\lambda^2}=\sin\theta$ (with $\sin\theta\ge0$ on $[0,\pi]$). A rotation by
$\theta$ raised to the $k$-th power is a rotation by $k\theta$, which is
the equation above after substituting $\cos(k\theta)=T_k(\lambda)$ and
$\sin(k\theta)=\sin\theta\,U_{k-1}(\lambda)=\sqrt{1-\lambda^2}\,U_{k-1}(\lambda)$
from the equation above.

**Example (The rotation picture with numbers: $\lambda=\tfrac12$).**

Take $\lambda=\tfrac12$, so $\theta=\arccos\tfrac12=60^{\circ}$ and
$\sqrt{1-\lambda^2}=\tfrac{\sqrt3}{2}$. Then
$$O(\tfrac12)=\begin{pmatrix}\tfrac12&-\tfrac{\sqrt3}{2}\\[2pt]
    \tfrac{\sqrt3}{2}&\tfrac12\end{pmatrix}
  \qquad(\text{rotation by }60^{\circ}),$$
and its powers are rotations by $k\cdot60^{\circ}$, whose $(1,1)$-entries read off the
Chebyshev values directly:
$$[O^2]_{11}=\cos120^{\circ}=-\tfrac12=T_2(\tfrac12),\quad
  [O^3]_{11}=\cos180^{\circ}=-1=T_3(\tfrac12),\quad
  [O^6]_{11}=\cos360^{\circ}=1=T_6(\tfrac12),$$
each matching the recurrence the equation above: $T_2(\tfrac12)=2(\tfrac12)^2-1
=-\tfrac12$ and $T_3(\tfrac12)=2\cdot\tfrac12\cdot(-\tfrac12)-\tfrac12=-1$. Raising the
single matrix $O=U(\lambda)Z$ to the $k$-th power *is* evaluating the degree-$k$
Chebyshev polynomial at $\lambda$---no polynomial arithmetic, just a rotation
iterated. Qubitization replaces $\tfrac12$ by an eigenvalue $\lambda_j$ of $A$ and
runs this one rotation on every eigenspace at once.

The content of that lemma is worth restating in plain terms: the
$(1,1)$-entry of $O(\lambda)^k$ is exactly $T_k(\lambda)$. If we could somehow
substitute a matrix $A$ for the scalar $\lambda$---replacing the $2\times2$ block by
a block encoding of $A$, and the reflection $Z$ by an appropriate reflection on the
ancilla---then the analogous power would be a block encoding of $T_k(A)$, with the
degree-$k$ Chebyshev polynomial computed at unit subnormalization and with no LCU
whatsoever. For a Hermitian $A=\sum_j\lambda_j\proj{v_j}$ the substitution is
literally the spectral one, $T_k(A)=\sum_j T_k(\lambda_j)\proj{v_j}$, applied
eigenvalue by eigenvalue. Making this precise is the business of qubitization.

## Qubitization of Hermitian matrices

Let $A$ be Hermitian with $\norm{A}\le1$ and spectral decomposition
$A=\sum_j\lambda_j\proj{v_j}$, and suppose first that we hold a *Hermitian*
block encoding $U_A\in\mathrm{HBE}_{1,m}(A)$ (that definition), i.e. $U_A=U_A\dg$ and $(\bra{0^m}\otimes\id)U_A(\ket{0^m}\otimes\id)=A$. Write
$\Pi\defeq\proj{0^m}\otimes\id$ for the projector onto the flagged subspace.

**Lemma (Invariant two-dimensional subspaces).**

For each eigenvector $\ket{v_j}$ of $A$ there is a normalized state
$\ket{\perp_j}$ with $\Pi\ket{\perp_j}=0$ such that
$\mathcal H_j\defeq\mathrm{span}\{\ket{0^m}\ket{v_j},\ket{\perp_j}\}$ is invariant
under $U_A$, and in the ordered basis $\mathcal B_j=\{\ket{0^m}\ket{v_j},
\ket{\perp_j}\}$
$$_{\mathcal B_j}
  =\begin{pmatrix}\lambda_j&\sqrt{1-\lambda_j^2}\\[2pt]
    \sqrt{1-\lambda_j^2}&-\lambda_j\end{pmatrix},
  \qquad
  [\Pi]_{\mathcal B_j}=\begin{pmatrix}1&0\\0&0\end{pmatrix}.$$

*Proof.*

Applying $U_A$ to the flagged eigenstate and using
$\Pi U_A\Pi=\proj{0^m}\otimes A$,
$$U_A\ket{0^m}\ket{v_j}=\ket{0^m}A\ket{v_j}+\ket{e_j^\perp}
  =\lambda_j\ket{0^m}\ket{v_j}+\ket{e_j^\perp},$$
where $\ket{e_j^\perp}\defeq(\id-\Pi)U_A\ket{0^m}\ket{v_j}$ satisfies
$\Pi\ket{e_j^\perp}=0$. Since $U_A$ is unitary the left side is a unit vector, so
$\norm{\ket{e_j^\perp}}^2=1-\lambda_j^2$ and we may write
$\ket{e_j^\perp}=\sqrt{1-\lambda_j^2}\,\ket{\perp_j}$ with $\ket{\perp_j}$
normalized (if $\lambda_j^2=1$ then $\ket{e_j^\perp}=0$ and $\mathcal H_j$ is
one-dimensional; take $\ket{\perp_j}$ to be any completing unit vector). This gives
the first column of the equation above. For the second, use $U_A=U_A\dg$: applying
$U_A$ to the equation above and $U_A^2=\id$,
$$\ket{0^m}\ket{v_j}=\lambda_j U_A\ket{0^m}\ket{v_j}
   +\sqrt{1-\lambda_j^2}\,U_A\ket{\perp_j}
  =\lambda_j^2\ket{0^m}\ket{v_j}+\lambda_j\sqrt{1-\lambda_j^2}\ket{\perp_j}
   +\sqrt{1-\lambda_j^2}\,U_A\ket{\perp_j},$$
so solving for $U_A\ket{\perp_j}$,
$$U_A\ket{\perp_j}=\sqrt{1-\lambda_j^2}\,\ket{0^m}\ket{v_j}-\lambda_j\ket{\perp_j},$$
which is the second column of the equation above. Hence $\mathcal H_j$ is invariant
and $U_A$ restricts to the stated reflection. The matrix of $\Pi$ is immediate since
$\Pi\ket{0^m}\ket{v_j}=\ket{0^m}\ket{v_j}$ and $\Pi\ket{\perp_j}=0$.

The name *qubitization* reflects the equation above: an arbitrarily large
block encoding $U_A$ is block-diagonalized into $N$ independent "qubits", one
$2\times2$ reflection per eigenvalue, and these blocks are precisely the reflection
$U(\lambda_j)$ of that lemma. Two eigenvectors with $\lambda_j^2=1$
require no companion state; all other subspaces are genuinely two-dimensional and
mutually orthogonal (the $\ket{\perp_j}$ are orthogonal because they live in
distinct eigenspaces of $A$ propagated by $U_A$).

To turn each reflection into the *rotation* $O(\lambda_j)$ of
the equation above we need the ancilla reflection $Z_\Pi\defeq2\Pi-\id$, which
acts as $+1$ on the flagged subspace and $-1$ on its complement, so
$[Z_\Pi]_{\mathcal B_j}=\mathrm{diag}(1,-1)=Z$. Then $O\defeq U_A Z_\Pi$ restricts to
$[O]_{\mathcal B_j}=[U_A]_{\mathcal B_j}\,Z=O(\lambda_j)$, exactly the rotation of
that lemma. This is the substitution promised in that section.

**Theorem (Qubitization: Chebyshev polynomials of a Hermitian matrix).**

Let $U_A\in\mathrm{HBE}_{1,m}(A)$ with $A$ Hermitian, $\norm{A}\le1$, and let
$Z_\Pi=2\Pi-\id$ with $\Pi=\proj{0^m}\otimes\id$. Then for every $k\ge0$,
$$\bigl(U_A Z_\Pi\bigr)^k\in\mathrm{BE}_{1,m}\bigl(T_k(A)\bigr),
  \qquad\text{i.e.}\qquad
  \Pi\,\bigl(U_A Z_\Pi\bigr)^k\,\Pi=\proj{0^m}\otimes T_k(A).$$

*Proof.*

Each $\mathcal H_j$ is invariant under both $U_A$ (that lemma) and
$Z_\Pi$ (which is diagonal in $\mathcal B_j$), hence under $O=U_A Z_\Pi$, and
$[O]_{\mathcal B_j}=O(\lambda_j)$. By that lemma,
$[O^k]_{\mathcal B_j}=O(\lambda_j)^k$ has $(1,1)$-entry $T_k(\lambda_j)$, i.e. $$\bra{0^m}\bra{v_j}\,O^k\,\ket{0^m}\ket{v_j}=T_k(\lambda_j).$$
For $j\ne j'$ the subspaces $\mathcal H_j,\mathcal H_{j'}$ are orthogonal, so
$\bra{0^m}\bra{v_{j'}}O^k\ket{0^m}\ket{v_j}=0$. The states
$\{\ket{0^m}\ket{v_j}\}_j$ form an orthonormal basis of the range of $\Pi$, so
the equation above says $\Pi O^k\Pi=\proj{0^m}\otimes\sum_j
T_k(\lambda_j)\proj{v_j}=\proj{0^m}\otimes T_k(A)$, which is
the equation above.

The reflection $Z_\Pi$ is cheap: for $m=1$ it is a single Pauli $Z$, and in general
it is a multiply-controlled phase implementable with one clean ancilla and an
$m$-fold Toffoli. One step of qubitization is therefore "$U_A$ then $Z_\Pi$", and
$k$ steps give $T_k(A)$ with a *single* query per Chebyshev degree and unit
subnormalization---compare the growing $\ell^1$ cost of building $T_k(A)$ as an LCU
of monomials in that chapter. Because $\{T_k\}_{k\ge0}$ is a basis of the
polynomials, one could now form any polynomial of $A$ by an LCU of the qubitized
Chebyshev blocks (that result); QSP will do the same job below with a
single ancilla and no LCU.

**Example (A degree-$2$ Chebyshev block).**

Take $A=\tfrac12\Xgate$ on one qubit, with the Hermitian block encoding
$U_A\in\mathrm{HBE}_{1,1}(A)$ of that theorem, so $m=1$ and
$Z_\Pi=\Zgate$ on the flag qubit. The two eigenvalues are $\lambda_\pm=\pm\tfrac12$
(eigenvectors $\ket\pm$), and that theorem gives
$(U_A\Zgate)^2\in\mathrm{BE}_{1,1}(T_2(A))$ with
$T_2(A)=2A^2-\id=2\cdot\tfrac14\id-\id=-\tfrac12\id$. One checks directly that
$T_2(\lambda_\pm)=2\cdot\tfrac14-1=-\tfrac12$, consistent with the spectral formula
$T_2(A)=\sum_\pm T_2(\lambda_\pm)\proj{\pm}=-\tfrac12\id$. Two queries to $U_A$ thus
block-encode $-\tfrac12\id$ exactly.

## Qubitization of general matrices

Hermitian block encodings of Hermitian matrices exist (that theorem) but may
demand extra structure and oracles. In general the block encoding $U_A$ of a
Hermitian $A$ is *not* itself Hermitian, and the single relation
the equation above no longer closes on a two-dimensional invariant subspace,
because we cannot re-apply $U_A$ and use $U_A^2=\id$. The remedy is to alternate
$U_A$ and $U_A\dg$. The result is that $U_A$ maps one two-dimensional subspace to a
*second* one, and $U_A\dg$ maps it back.

**Lemma (Qubitization with a general block encoding).**

Let $U_A\in\mathrm{BE}_{1,m}(A)$ with $A$ Hermitian, $\norm{A}\le1$, and
$\Pi=\proj{0^m}\otimes\id$. For each eigenvector $\ket{v_j}$ there are normalized
states $\ket{\perp_j},\ket{\perp_j'}$ (both annihilated by $\Pi$) such that, with
$\mathcal B_j=\{\ket{0^m}\ket{v_j},\ket{\perp_j}\}$ and
$\mathcal B_j'=\{\ket{0^m}\ket{v_j},\ket{\perp_j'}\}$,
$$^{\mathcal B_j'}_{\mathcal B_j}
  =\begin{pmatrix}\lambda_j&\sqrt{1-\lambda_j^2}\\[2pt]
    \sqrt{1-\lambda_j^2}&-\lambda_j\end{pmatrix}
  =[U_A\dg]^{\mathcal B_j}_{\mathcal B_j'}.$$
Consequently $\mathcal H_j=\mathrm{span}\,\mathcal B_j$ is invariant under
$\widetilde O\defeq U_A\dg Z_\Pi U_A Z_\Pi$, with
$$_{\mathcal B_j}
  =\begin{pmatrix}\lambda_j&-\sqrt{1-\lambda_j^2}\\[2pt]
    \sqrt{1-\lambda_j^2}&\lambda_j\end{pmatrix}^{\!2}
  =O(\lambda_j)^2 .$$

*Proof.*

As in the equation above, $U_A\ket{0^m}\ket{v_j}=\lambda_j\ket{0^m}\ket{v_j}
+\sqrt{1-\lambda_j^2}\,\ket{\perp_j'}$ defines the normalized $\ket{\perp_j'}$; this
is the first column of the equation above. Because $A$ is Hermitian, $U_A\dg$ is
also a block encoding of $A$ ($\Pi U_A\dg\Pi=(\Pi U_A\Pi)\dg=\proj{0^m}\otimes A$),
so likewise $U_A\dg\ket{0^m}\ket{v_j}=\lambda_j\ket{0^m}\ket{v_j}
+\sqrt{1-\lambda_j^2}\,\ket{\perp_j}$ defines $\ket{\perp_j}$. Applying $U_A$ to this
and using $U_A U_A\dg=\id$ gives, exactly as in the equation above--
the equation above, $U_A\ket{\perp_j}=\sqrt{1-\lambda_j^2}\,\ket{0^m}\ket{v_j}
-\lambda_j\ket{\perp_j'}$, the second column of the equation above; the identity
$[U_A\dg]^{\mathcal B_j}_{\mathcal B_j'}=([U_A]^{\mathcal B_j'}_{\mathcal B_j})\dg$
equals the same real symmetric matrix. Thus $U_A$ carries
$\mathcal H_j\to\mathcal H_j'$ and $U_A\dg$ carries $\mathcal H_j'\to\mathcal H_j$;
each of $Z_\Pi$ fixes the shared vector $\ket{0^m}\ket{v_j}$ and flips the
perpendicular one. Composing the four factors of $\widetilde O=U_A\dg Z_\Pi U_A Z_\Pi$ maps
$\mathcal H_j\to\mathcal H_j$: the inner $Z_\Pi U_A Z_\Pi$ carries $\mathcal H_j$ to
$\mathcal H_j'$ acting as $Z\,[U_A]^{\mathcal B_j'}_{\mathcal B_j}Z$ up to the shared
vector, and the outer $U_A\dg$ returns to $\mathcal H_j$. In the basis $\mathcal B_j$
the composite matrix is $\bigl([U_A]^{\mathcal B_j'}_{\mathcal B_j}Z\bigr)^2
=\bigl(U(\lambda_j)Z\bigr)^2=O(\lambda_j)^2$, which is
the equation above.

**Theorem (Chebyshev polynomials from a general block encoding).**

With the notation of that lemma, for every $k\ge1$,
$$\bigl(U_A\dg Z_\Pi U_A Z_\Pi\bigr)^k\in\mathrm{BE}_{1,m}\bigl(T_{2k}(A)\bigr),
  \qquad
  U_A Z_\Pi\bigl(U_A\dg Z_\Pi U_A Z_\Pi\bigr)^k\in\mathrm{BE}_{1,m}\bigl(T_{2k+1}(A)\bigr).$$
Hence the even Chebyshev polynomial $T_{2k}(A)$ is obtained by $k$ alternating pairs
$U_A\dg Z_\Pi,\,U_A Z_\Pi$, and the odd one $T_{2k+1}(A)$ by one extra $U_A Z_\Pi$.

*Proof.*

By the equation above, $[\widetilde O^k]_{\mathcal B_j}=O(\lambda_j)^{2k}$, whose
$(1,1)$-entry is $T_{2k}(\lambda_j)$ (that lemma); orthogonality of the
$\mathcal H_j$ and completeness of $\{\ket{0^m}\ket{v_j}\}$ then give
$\Pi\widetilde O^k\Pi=\proj{0^m}\otimes T_{2k}(A)$ as in that theorem. For
the odd case, one more $U_A Z_\Pi$ sends $\mathcal H_j\to\mathcal H_j'$; since the two
bases share the vector $\ket{0^m}\ket{v_j}$, the surviving flagged block is the
$(1,1)$-entry of $O(\lambda_j)^{2k+1}$, namely $T_{2k+1}(\lambda_j)$, giving the
second encoding in the equation above.

The lesson is that a non-Hermitian block encoding produces Chebyshev polynomials of
$A$ just as before, but the fundamental building block is a $U_A,U_A\dg$
*pair*, and even/odd degrees must be assembled separately. This alternation is
not an artifact: it is the same structural feature that forces the parity condition
in the QSP theorem, to which we now turn. (When $U_A$ happens to be Hermitian,
$U_A\dg=U_A$ and that theorem collapses to that theorem.)

## Quantum signal processing

Chebyshev polynomials are only the phase-free special case. Interleaving
single-qubit $Z$-rotations $\ee^{\ii\phi_j Z}$ between the qubitization steps
produces a far richer family. Because every operation preserves the
two-dimensional subspaces $\mathcal H_j$, the entire question reduces, by
qubitization, to a statement about $\mathrm{SU}(2)$: what $2\times2$ matrix
polynomials arise from a product of the fixed rotation $O(\lambda)$ of
the equation above and free $Z$-rotations? This is *quantum signal
processing*.

**Definition (QSP sequence).**

For phase factors $\Phi=(\phi_0,\phi_1,\dots,\phi_d)\in\R^{d+1}$ and
$\lambda\in[-1,1]$, define the *QSP unitary*
$$U_\Phi(\lambda)\defeq\ee^{\ii\phi_0 Z}\prod_{j=1}^{d}
     \Bigl[O(\lambda)\,\ee^{\ii\phi_j Z}\Bigr],
  \qquad O(\lambda)=\begin{pmatrix}\lambda&-\sqrt{1-\lambda^2}\\[2pt]
    \sqrt{1-\lambda^2}&\lambda\end{pmatrix}.$$

The number of signal operators $O(\lambda)$ is $d$, the "query depth". The
following theorem characterizes exactly the polynomials that appear in the top-left
entry.

**Theorem (Quantum signal processing).**

There exist phase factors $\Phi\in\R^{d+1}$ with
$$U_\Phi(\lambda)
  =\begin{pmatrix}P(\lambda)&-Q(\lambda)\sqrt{1-\lambda^2}\\[3pt]
    Q^*(\lambda)\sqrt{1-\lambda^2}&P^*(\lambda)\end{pmatrix}$$
if and only if $P,Q\in\C[\lambda]$ satisfy

1.  $\deg P\le d$ and $\deg Q\le d-1$;

2.  $P$ has parity $d\bmod2$ and $Q$ has parity $(d-1)\bmod2$;

3.  $\abs{P(\lambda)}^2+(1-\lambda^2)\abs{Q(\lambda)}^2=1$ for all
    $\lambda\in[-1,1]$.

Here $\deg Q=-1$ means $Q\equiv0$, and $P^*,Q^*$ denote coefficientwise complex
conjugation.

*Proof (Proof of the forward direction ($\Rightarrow$)).*

We prove by induction on $d$ that $U_\Phi(\lambda)$ has the form
the equation above with $P,Q$ satisfying (1)--(2); condition (3) then follows
for free, since $U_\Phi(\lambda)$ is a product of unitaries, hence unitary, and
comparing $\abs{\cdot}^2$ of its first column gives
$\abs{P}^2+(1-\lambda^2)\abs{Q}^2=1$.

*Base case $d=0$.* Here $U_\Phi(\lambda)=\ee^{\ii\phi_0 Z}
=\mathrm{diag}(\ee^{\ii\phi_0},\ee^{-\ii\phi_0})$, so $P(\lambda)=\ee^{\ii\phi_0}$
(degree $0$, even) and $Q\equiv0$, satisfying (1)--(2).

*Inductive step.* Suppose the length-$(d-1)$ product
$U_{(\phi_0,\dots,\phi_{d-1})}(\lambda)$ equals the equation above with
$P,Q$ of degrees $\le d-1,\le d-2$ and parities $(d-1)\bmod2,(d-2)\bmod2$. Appending
one more block multiplies on the right by $O(\lambda)\,\ee^{\ii\phi_d Z}$:
$$\begin{aligned}
  \begin{pmatrix}P&-Q\sqrt{1-\lambda^2}\\ Q^*\sqrt{1-\lambda^2}&P^*\end{pmatrix}
  \begin{pmatrix}\lambda&-\sqrt{1-\lambda^2}\\ \sqrt{1-\lambda^2}&\lambda\end{pmatrix}
  \begin{pmatrix}\ee^{\ii\phi_d}&0\\0&\ee^{-\ii\phi_d}\end{pmatrix}.
  
\end{aligned}$$
Carrying out the first product, the new top-left entry (before the phase) is
$\lambda P-(1-\lambda^2)Q$ and the new top-right entry is
$-\sqrt{1-\lambda^2}\,(P+\lambda Q)$; after the diagonal phase the matrix is
$$\begin{pmatrix}
    \ee^{\ii\phi_d}\bigl(\lambda P-(1-\lambda^2)Q\bigr) &
    -\ee^{-\ii\phi_d}\sqrt{1-\lambda^2}\,(P+\lambda Q)\\[4pt]
    \ee^{\ii\phi_d}\sqrt{1-\lambda^2}\,(P^*+\lambda Q^*)&
    \ee^{-\ii\phi_d}\bigl(\lambda P^*-(1-\lambda^2)Q^*\bigr)
  \end{pmatrix}.$$
This has the form the equation above with
$$\widetilde P=\ee^{\ii\phi_d}\bigl(\lambda P-(1-\lambda^2)Q\bigr),\qquad
  \widetilde Q=\ee^{-\ii\phi_d}(P+\lambda Q).$$
Degrees: $\deg\widetilde P\le\max(1+\deg P,2+\deg Q)\le d$ and
$\deg\widetilde Q\le\max(\deg P,1+\deg Q)\le d-1$, giving (1). Parities: multiplying
by $\lambda$ or $(1-\lambda^2)\cdot(\text{something of parity }d-2)$ flips or
preserves parity so that $\widetilde P$ has parity $d\bmod2$ and $\widetilde Q$ has
parity $(d-1)\bmod2$, giving (2). This completes the induction, and (3) follows from
unitarity as noted.

The converse---that *every* $P,Q$ satisfying (1)--(3) is realized by some
$\Phi$---is proved by the reverse induction: given $P,Q$ of degree $\ell$, condition
(3) forces the two leading coefficients to have equal modulus, and a single phase
$\ee^{2\ii\phi_d}=(\text{ratio of leading coefficients})$ peels off one degree,
reducing to a shorter QSP problem; iterating recovers all $\phi_j$. We refer to the
reading for the details (it is essentially the Gilyén--Su--Low--Wiebe argument),
and record the practically important consequences.

**Corollary (QSP for real target polynomials).**

Let $p\in\R[\lambda]$ have $\deg p=d$, parity $d\bmod2$, and $\abs{p(\lambda)}\le1$
on $[-1,1]$. Then there exist $P,Q$ satisfying [(1)--(3)]{.upright} of
that theorem with $\operatorname{Re}P=p$, and correspondingly phase factors
$\Phi$ with $\operatorname{Re}\!\bigl[U_\Phi(\lambda)_{11}\bigr]=p(\lambda)$.

The corollary is what one uses in practice: almost every application (matrix
inversion, Hamiltonian simulation, eigenvalue filtering) starts from a real target
polynomial $p$, scales it so that $\abs{p}\le1$, and then produces $P$ with
$\operatorname{Re}P=p$. The real part is extracted by one extra Hadamard-sandwiched
ancilla, exactly the "$P_{\mathrm{Re}}$" construction of qubitization: running
$U_\Phi$ and $U_{-\Phi}$ (which block-encodes $P^*$) in an LCU with equal weights
gives $\tfrac12(P+P^*)=\operatorname{Re}P=p$. Lifting from the scalar $\lambda$ to a
Hermitian $A$ is now automatic: replace $O(\lambda)$ by $O=U_A Z_\Pi$ (Hermitian
encoding) or by the alternating $U_A,U_A\dg$ pattern (general encoding), and each
$\ee^{\ii\phi_j Z}$ by the controlled rotation $\ee^{\ii\phi_j Z_\Pi}$ on the flag;
since every factor preserves the subspaces $\mathcal H_j$, the flagged block is
$P(A)=\sum_j P(\lambda_j)\proj{v_j}$. Thus $d$ queries to $U_A$ block-encode $P(A)$
with unit subnormalization. This is the *quantum eigenvalue transformation*;
its singular-value extension to non-Hermitian $A$ is QSVT (that chapter).

**Example (QSP realizations of low-degree polynomials).**

*(a) $T_d$ via all-zero phases.* Setting $\Phi=(0,\dots,0)$ in
the equation above gives $U_\Phi(\lambda)=O(\lambda)^d$, whose $(1,1)$-entry is
$T_d(\lambda)$ by that lemma. Thus $P=T_d$, $Q=U_{d-1}$: the Chebyshev
polynomials are the phase-free QSP polynomials, consistent with
that theorem.

*(b) Degree $1$.* For $d=1$,
$U_\Phi(\lambda)=\ee^{\ii\phi_0 Z}O(\lambda)\ee^{\ii\phi_1 Z}$ has top-left entry
$$P(\lambda)=\ee^{\ii(\phi_0+\phi_1)}\lambda,$$
so $\Phi=(0,0)$ realizes $P(\lambda)=\lambda=T_1(\lambda)$ (real, odd), while
$\Phi=(\tfrac\pi4,\tfrac\pi4)$ realizes $P(\lambda)=\ii\lambda$; in every case
$\abs{P(\lambda)}=\abs\lambda\le1$ and $\deg P=1$, matching (1)--(3) with
$Q=\ee^{\ii(\phi_1-\phi_0)}$ a constant.

*(c) Degree $2$, target $T_2$, symmetric phases.* A more symmetric choice
realizing $T_2(\lambda)=2\lambda^2-1$ uses the eigenvalue-transform convention
$U_A\ee^{\ii e_j Z}$ (with $U_A$ the reflection, not $O$): the phases
$e=(\tfrac\pi4,\tfrac\pi2,\tfrac\pi4)$ yield $(-\ii)^2 U_{e,11}=2\lambda^2-1=T_2$.
Direct multiplication of $\ee^{\ii\frac\pi4 Z}U_A\ee^{\ii\frac\pi2 Z}U_A
\ee^{\ii\frac\pi4 Z}$ confirms the top-left entry equals $\ii\,(2\lambda^2-1)$, whose
$(-\ii)^2=-1$ times the leading sign returns $T_2$; equivalently in the
$O$-convention $\Phi=(0,0,0)$ gives $O(\lambda)^2$ with $(1,1)$-entry
$2\lambda^2-1$. These small cases are the ones we verify by hand (and, in the
companion notebook, numerically: the QSP unitary equals the claimed polynomial to
machine precision).

## Finding phase factors

QSP is only useful if the phase factors $\Phi$ realizing a desired $p$ can actually
be computed. When QSP was introduced this was considered a serious obstacle; it is
now, for all practical purposes, solved. We survey the three main families of
methods; all take as input a real target $p$ of degree $d$, parity $d\bmod2$, with
$\norm{p}_\infty<1$ on $[-1,1]$, and return $\Phi\in\R^{d+1}$ (of which only
$\lceil(d+1)/2\rceil$ are free, by parity).

##### Direct (root-finding) methods.

The converse proof of
that theorem is constructive: given $P,Q$ one strips one degree at a time,
each step reading off a phase from the ratio of leading coefficients. The missing
ingredient is $Q$, obtained from the *completion* step---given $P$ with
$\abs{P}\le1$, find $Q$ with $\abs{P}^2+(1-\lambda^2)\abs{Q}^2=1$ by factoring the
degree-$2d$ polynomial $1-\abs{P(\lambda)}^2$ over its roots and grouping them into a
Fejér--Riesz square. This is the Remez-exchange / root-finding approach: exact in
principle, but the root-finding is numerically delicate in high precision because
the roots of $1-P^2$ cluster and the extraction is ill-conditioned, requiring
high-precision arithmetic for large $d$.

##### Optimization-based methods.

Rather than solve for $\Phi$ exactly, one
minimizes the least-squares mismatch between the QSP output and the target on a
Chebyshev grid. Fixing the parity, only $\widetilde d=\lceil(d+1)/2\rceil$ phases are
free, and $p$ is determined by its values at $\widetilde d$ points, which one takes
to be the positive Chebyshev nodes $\lambda_k=\cos\!\bigl(\tfrac{(2k-1)\pi}
{4\widetilde d}\bigr)$. Minimizing
$$F(\Phi)=\frac{1}{\widetilde d}\sum_{k=1}^{\widetilde d}
     \Bigl(\operatorname{Re}\bigl[U_\Phi(\lambda_k)_{11}\bigr]-p(\lambda_k)\Bigr)^2$$
over $\Phi$ drives $F$ to $0$ at a valid phase set. The remarkable empirical (and
now proven) fact is that starting from the symmetric ansatz
$\Phi_0=(\tfrac\pi4,0,\dots,0,\tfrac\pi4)$ this nonconvex problem behaves almost
convexly near the solution, so quasi-Newton descent converges rapidly and stably
even in double precision for degrees in the thousands.

##### Symmetric QSP and fixed-point iteration.

Restricting to
*symmetric* phase sequences ($\phi_j=\phi_{d-j}$) makes $U_\Phi(\lambda)_{11}$
automatically real with the correct parity, halving the unknowns and turning the
completion problem into a well-conditioned fixed-point / Newton iteration on the
phases directly (no explicit $Q$ needed). Related developments recast phase-factor
finding as a nonlinear Fourier transform, for which fast and provably stable
algorithms exist. In practice one calls a mature routine (e.g. the
`QSPPACK` optimization solver or a symmetric-QSP Newton solver) as a black
box: supply the coefficients of $p$, receive $\Phi$, and the QSP circuit of
that section then block-encodes $p(A)$ with $d$ queries.

The upshot for the rest of Part II is that we may treat "block-encode $p(A)$ for a
bounded degree-$d$ polynomial $p$" as a primitive costing $\bigo(d)$ queries to
$U_A$, one extra ancilla, and $d$ classically-precomputed single-qubit rotations,
with phase factors supplied by one of the solvers above. This primitive, and its
singular-value form, is the engine of quantum linear algebra: matrix inversion,
Hamiltonian simulation, and eigenvalue filtering are all instances of choosing the
right $p$, as that chapter will show.
